# English Indices of Deprivation

The dataset analyzed in this notebook presents the survey results of the 2025 English Indices of Deprivation survey. This survey "measure[s] relative levels of deprivation in 33,755 small areas or neighbourhoods, called Lower-layer Super Output Areas (LSOAs), in England."

Retrieved From: https://deprivation.communities.gov.uk/download-all

The Government of the United Kingdom defines deprivation as "people’s unmet needs, a lack of access to opportunities and resources which we might expect in our society." REFERENCE

The English Indices of Deprivation survey measures levels of deprivation along 7 where each domain is weighted differently to determine the Index of Multiple Deprivation (IMD) for each LSOA:
- Income Deprivation (22.5)
- Employment Deprivation (22.5)
- Education, Skills and Training Deprivation (13.5)
- Health and Disability Deprivation (13.5)
- Crime Deprivation (9.3)
- Barriers to Housing and Services Deprivation (9.3)
- Living Environment Deprivation (9.3)
