# Lower Layer Super Output Areas

The Lower Layer Super Output Areas data contains a micro breakdown of London, specifically a neighbourhood breakdown. In this notebook I inspect the data and more importantly filter for Tower Hamlets areas.

Retrieved From: https://geoportal.statistics.gov.uk/datasets/ons::lower-layer-super-output-areas-december-2021-boundaries-ew-bsc-v4-2/about

```python
import pandas as pd
import json
```

## CSV LSOA

```python
df = pd.read_csv('../data/raw_data/lsoa/Lower_layer_Super_Output_Areas_December_2021_Boundaries_EW_BSC_V4_3901388190129020682.csv')
```

As we can see there are 169 LSOA codes in Tower Hamlets, now I will import the GeoJSON version of this dataset to confirm that there are indeed 169 in that file which will be used to create the map in the app.

## GeoJSON LSOA

```python
with open('../data/raw_data/lsoa/Lower_layer_Super_Output_Areas_December_2021_Boundaries_EW_BSC_V4_-4299016806856585929.geojson', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
tower_hamlets_only = [
    feature for feature in data['features']
    if 'Tower Hamlets' in feature['properties'].get('LSOA21NM', '')
]
```

```python
filtered_data = {
    "type": "FeatureCollection",
    "name": "Tower_Hamlets_LSOA",
    "crs": data.get("crs"),
    "features": tower_hamlets_only
}
```

Exporting the Tower Hamlets-specific GeoJSON.

```python
with open('../data/clean_data/lsoa_clean.geojson', 'w', encoding='utf-8') as f:
    json.dump(filtered_data, f)
```

#### Inspecting the filtered GeoJSON file

```python
with open('../data/clean_data/lsoa_clean.geojson', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
print(f"Number of rows: {len(data['features'])}")
```
