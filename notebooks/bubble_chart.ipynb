## A bubble chart of lsoa income vs deprivation

Each bubble will be the number of charities in the activity focus.

```python
import json
import pandas as pd
import plotly.express as px
```

```python
with open("../charity_watch_streamlit/data/charities_with_deprivation.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
df = pd.DataFrame(data)
df["income"] = pd.to_numeric(df["income"], errors="coerce")
df["imdScore"] = pd.to_numeric(df["imdScore"], errors="coerce")
```

```python
agg = (df.groupby("primaryFocus", dropna=False).agg(avg_income=("income", "mean"),avg_imd=("imdScore", "mean"), charity_count=("name", "count"),).reset_index())
```

```python
fig = px.scatter(
    agg,
    x="avg_imd",
    y="avg_income",
    size="charity_count",
    color="primaryFocus",
    hover_name="primaryFocus",
    size_max=70,
    title="Primary Focus: Income vs Deprivation",
)

fig.update_layout(
    xaxis_title="Average IMD Score",
    yaxis_title="Average Income",
)

fig.show()
```
