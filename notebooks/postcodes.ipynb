# Office for National Statistics Postcodes Dataset

This dataset contains all postcodes in England and Wales. This dataset is fundamental in order to join the charities dataset with the english indices of deprivation dataset. As stated this dataset contains postcodes which is a field in the charities data and it also contains LSOA 2021 codes which is a field in the deprivation dataset.

```python
import pandas as pd
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../data/raw_data/ONSPD_MAY_2025_UK_E.csv')
```

The local authority code for Tower Hamlets is E09000030, so we can filter by that.

```python
pc_tower_hamlets = df[df['oslaua'] == 'E09000030']
```
